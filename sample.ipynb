```python
# ---------------------------------------------------------
# 0. Libs
# ---------------------------------------------------------
import os, warnings, pandas as pd, matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from neuralforecast.core import NeuralForecast
from neuralforecast.models import LSTM, TimesNet, NHITS
import pytorch_lightning as pl, torch, numpy as np

warnings.filterwarnings("ignore")
os.makedirs("results", exist_ok=True)

# Reproductibilité
pl.seed_everything(1, workers=True)
torch.manual_seed(1); np.random.seed(1)

# ---------------------------------------------------------
# 1. Data
# ---------------------------------------------------------
df = pd.read_csv("weather.csv")
df["date"] = pd.to_datetime(df["date"])
df = df.set_index("date").sort_index()

# Ici on prend une fréquence horaire (si tu veux 10 min -> "10min")
df = df.resample("10min").mean().interpolate("time")

# NeuralForecast attend les colonnes: unique_id, ds, y
data = pd.DataFrame({
    "unique_id": "weather",
    "ds": df.index,
    "y": df["T (degC)"]
})

# Split train/test
H = 24                  # horizon = 24 heures
TEST_SIZE = H * 7       # 1 semaine pour test
train = data.iloc[:-TEST_SIZE].copy()
test  = data.iloc[-TEST_SIZE:].copy()

# ---------------------------------------------------------
# 2. Modèles
# ---------------------------------------------------------
models = [
    LSTM(    h=H, input_size=2*H, max_steps=200, val_check_steps=20),
    TimesNet(h=H, input_size=2*H, max_steps=200, val_check_steps=20),
    NHITS(   h=H, input_size=2*H, max_steps=200, val_check_steps=20),
]

nf = NeuralForecast(models=models, freq="H")
nf.fit(df=train)

# ---------------------------------------------------------
# 3. Prédictions
# ---------------------------------------------------------
preds = nf.predict(df=data).reset_index()

# On garde les 24 dernières heures pour comparaison
y_true = test.tail(H).reset_index(drop=True)
y_pred = preds.tail(H).reset_index(drop=True)

# Fusion par position
plot_df = pd.concat([y_true["y"], y_pred[["LSTM","TimesNet","NHITS"]]], axis=1)
plot_df.columns = ["y","LSTM","TimesNet","NHITS"]

# ---------------------------------------------------------
# 4. Métriques
# ---------------------------------------------------------
metrics = []
for m in ["LSTM", "TimesNet", "NHITS"]:
    mae = mean_absolute_error(plot_df["y"], plot_df[m])
    rmse = np.sqrt(mean_squared_error(plot_df["y"], plot_df[m]))  # ✅ compatible partout
    metrics.append({"model": m, "MAE": mae, "RMSE": rmse})

metrics_df = pd.DataFrame(metrics)
metrics_df.to_csv("results/metrics1.csv", index=False)
print("\n📊 Metrics :\n", metrics_df)

# ---------------------------------------------------------
# 5. Visualisation
# ---------------------------------------------------------
plt.figure(figsize=(14,4))
plt.plot(plot_df.index, plot_df["y"], label="True", color="black", lw=2)
for m in ["LSTM", "TimesNet", "NHITS"]:
    plt.plot(plot_df.index, plot_df[m], label=m)
plt.xticks(rotation=45); plt.grid(alpha=.3)
plt.title("Prévision 24 h – pas = 1h")
plt.ylabel("Température (°C)")
plt.legend()
plt.tight_layout()
plt.savefig("results/forecast1.png", dpi=300)
plt.show()

print("\n✅ Fichiers créés : results/metrics1.csv  &  results/forecast1.png")
```

```python
# =========================================================
# 0) Libs
# =========================================================
import os, warnings, numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from neuralforecast.core import NeuralForecast
from neuralforecast.models import LSTM, TimesNet, NHITS
import pytorch_lightning as pl, torch

warnings.filterwarnings("ignore")
os.makedirs("results", exist_ok=True)

# Seed
SEED = 1
pl.seed_everything(SEED, workers=True)
torch.manual_seed(SEED); np.random.seed(SEED)

# Détection GPU pour Lightning
ACCEL = "gpu" if torch.cuda.is_available() else "cpu"
DEVICES = 1

TRAINER_KW = dict(
    accelerator=ACCEL,
    devices=DEVICES,
    enable_progress_bar=True,
    inference_mode=True,
    logger=False
)

# =========================================================
# 1) Données
#    - on resample à 10 min pour coller à ton pipeline
#    - on utilise T (degC) comme cible y
#    - toutes les autres colonnes numériques = exogènes historiques
# =========================================================
raw = pd.read_csv("weather.csv")
raw["date"] = pd.to_datetime(raw["date"])
raw = raw.set_index("date").sort_index()

# resample 10 minutes + interpolation temporelle
df10 = raw.resample("10min").mean().interpolate("time")

# features numériques
num_cols = df10.select_dtypes(include=["number"]).columns.tolist()
assert "T (degC)" in num_cols, "Colonne cible 'T (degC)' introuvable."
y_col = "T (degC)"

# exogènes historiques = toutes les numériques sauf la cible
hist_cols = [c for c in num_cols if c != y_col]

# cadre NeuralForecast
data = pd.DataFrame({
    "unique_id": "weather",
    "ds": df10.index,
    "y": df10[y_col]
}).reset_index(drop=True)

# on rajoute les exogènes au même DataFrame (obligatoire pour NF)
data = pd.concat([data, df10[hist_cols].reset_index(drop=True)], axis=1)

# Split
H = 24 * 6             # 24 h à 10 min
TEST_SIZE = H * 7      # 1 semaine pour test
train = data.iloc[:-TEST_SIZE].copy()
test  = data.iloc[-TEST_SIZE:].copy()

# =========================================================
# 2) Modèles
#    - LSTM & NHITS : avec exogènes historiques
#    - TimesNet : sans hist_exog (limitation du modèle)
# =========================================================
# =========================================================
# 2) Modèles
# =========================================================
common = dict(
    h=H,
    input_size=7*24*6,          # ~1 semaine d'historique
    step_size=H,
    learning_rate=1e-3,
    batch_size=32,
    valid_batch_size=64,
    windows_batch_size=128,
    val_check_steps=50,
    early_stop_patience_steps=400,
    scaler_type="robust",
    random_seed=SEED,
    **TRAINER_KW                # ✅ on déplie le dict
)

mdl_lstm = LSTM(**common, hist_exog_list=hist_cols, max_steps=600)
mdl_timesnet = TimesNet(**common, max_steps=1000)  # ⚠️ sans hist_exog
mdl_nhits = NHITS(**common, hist_exog_list=hist_cols, max_steps=1000)

models = [mdl_lstm, mdl_timesnet, mdl_nhits]
nf = NeuralForecast(models=models, freq="10min")

nf.fit(df=train, val_size=3*H)  # ✅ fonctionne

# =========================================================
# 3) Entraînement
#    - val_size = 3 horizons pour early stopping
# =========================================================
nf = NeuralForecast(models=models, freq="10min")
nf.fit(df=train, val_size=3*H)

# =========================================================
# 4) Prédictions H pas (24h) et alignement
# =========================================================
preds = nf.predict(df=data).reset_index()

cols_models = [c for c in preds.columns if c in ["LSTM","TimesNet","NHITS"]]

# Si 'ds' est présent dans les colonnes → on aligne sur les vraies dates
if "ds" in preds.columns:
    y_true = test.tail(H)[["ds", "y"]].reset_index(drop=True)
    y_pred = preds.tail(H)[["ds"] + cols_models].reset_index(drop=True)
    plot_df = y_true.merge(y_pred, on="ds", how="inner")
else:
    # sinon fallback par position
    print("⚠️ Pas de colonne 'ds' dans preds → alignement par index")
    y_true = test.tail(H).reset_index(drop=True)
    y_pred = preds.tail(H).reset_index(drop=True)
    plot_df = pd.concat([y_true["y"], y_pred[cols_models]], axis=1)
    plot_df.columns = ["y"] + cols_models

# Vérif de sécurité
if len(plot_df) != H:
    print(f"⚠️ Alignement: attendu {H}, obtenu {len(plot_df)}")


# =========================================================
# 5) Métriques (RMSE de façon compatible)
# =========================================================
metrics = []
for m in ["LSTM","TimesNet","NHITS"]:
    if m in plot_df.columns:
        mae  = mean_absolute_error(plot_df["y"], plot_df[m])
        rmse = np.sqrt(mean_squared_error(plot_df["y"], plot_df[m]))  # robuste
        metrics.append({"model": m, "MAE": mae, "RMSE": rmse})

metrics_df = pd.DataFrame(metrics).sort_values("RMSE")
metrics_df.to_csv("results/metrics1.csv", index=False)
print("\n📊 Metrics (24h):\n", metrics_df)

# =========================================================
# 6) Visualisation
# =========================================================
plt.figure(figsize=(14,4))
plt.plot(plot_df["ds"], plot_df["y"], label="True", color="black", lw=2)
for m in ["LSTM","TimesNet","NHITS"]:
    if m in plot_df.columns:
        plt.plot(plot_df["ds"], plot_df[m], label=m)
plt.xticks(rotation=45); plt.grid(alpha=.3)
plt.title("Prévision 24 h – pas = 10 min")
plt.ylabel("Température (°C)")
plt.legend()
plt.tight_layout()
plt.savefig("results/forecast1.png", dpi=300)
plt.show()

print("\n✅ Fichiers créés : results/metrics1.csv  &  results/forecast1.png")

# (Optionnel) Sauvegarde des modèles pour réutilisation ultérieure
# nf.save(path="./checkpoints/weather_run/", overwrite=True, save_dataset=True)
```

```python
# =========================================================
# 0) Imports & style
# =========================================================
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.animation import FuncAnimation, FFMpegWriter
from sklearn.metrics import mean_absolute_error, mean_squared_error

os.makedirs("results", exist_ok=True)
sns.set_theme(style="whitegrid", context="talk")

# =========================================================
# 1) Chargement des données (2 options)
#    A) depuis CSV exportés
#    B) depuis variables en mémoire (décommente si besoin)
# =========================================================
# ---- A) depuis CSV
pred_path = "results/predictions.csv"     # colonnes attendues: ds, y, [LSTM, TimesNet, NHITS]
metrics_path = "results/metrics1.csv"     # colonnes: model, MAE, RMSE

pred_df = pd.read_csv(pred_path, parse_dates=["ds"])
metrics_df = pd.read_csv(metrics_path)

# ---- B) depuis variables en mémoire (ex: plot_df ou y_true/preds)
# # Si tu as `plot_df` avec colonnes: ds, y, LSTM, TimesNet, NHITS
# pred_df = plot_df.copy()
# # Si tu n'as pas ds dans plot_df, fabrique-le depuis l'index temps
# # pred_df["ds"] = ton_index_time_correspondant

# colonnes modèles présentes
model_cols = [c for c in ["LSTM","TimesNet","NHITS"] if c in pred_df.columns]

# Sécurité: trier par temps
pred_df = pred_df.sort_values("ds").reset_index(drop=True)

# =========================================================
# 2) Barplots MAE / RMSE
# =========================================================
if {"model","MAE","RMSE"}.issubset(metrics_df.columns):
    fig, ax = plt.subplots(1,2, figsize=(14,5), sharey=False)
    sns.barplot(data=metrics_df, x="model", y="MAE", ax=ax[0])
    ax[0].set_title("MAE par modèle"); ax[0].set_xlabel(""); ax[0].set_ylabel("MAE")

    sns.barplot(data=metrics_df, x="model", y="RMSE", ax=ax[1])
    ax[1].set_title("RMSE par modèle"); ax[1].set_xlabel(""); ax[1].set_ylabel("RMSE")

    plt.tight_layout()
    plt.savefig("results/model_comparison.png", dpi=200)
    plt.show()
    print("✅ model_comparison.png")

# =========================================================
# 3) Dernier horizon H : True vs Modèles (graphique de démonstration)
#    -> si tu connais H, mets-le; sinon on prend les N derniers points
# =========================================================
H = 144  # par ex. 24h à pas 10 min; ajuste selon ton pas
lastH = pred_df.tail(min(H, len(pred_df))).copy()

fig, ax = plt.subplots(figsize=(14,4))
ax.plot(lastH["ds"], lastH["y"], label="True", lw=2, color="black")
for m in model_cols:
    ax.plot(lastH["ds"], lastH[m], label=m, alpha=.9)
ax.set_title(f"Prévision – Dernier horizon (H={len(lastH)})")
ax.set_xlabel("Temps"); ax.set_ylabel("y")
ax.legend(ncol=len(model_cols)+1)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("results/forecast_lastH.png", dpi=200)
plt.show()
print("✅ forecast_lastH.png")

# =========================================================
# 4) Distribution des résidus (erreurs)
# =========================================================
resid = []
for m in model_cols:
    r = pred_df["y"].values - pred_df[m].values
    resid.append(pd.DataFrame({"model": m, "residual": r}))
resid_df = pd.concat(resid, ignore_index=True)

plt.figure(figsize=(12,5))
sns.kdeplot(data=resid_df, x="residual", hue="model", fill=True, common_norm=False, alpha=.25)
plt.title("Distribution des résidus (y - y_hat)")
plt.xlabel("Erreur"); plt.ylabel("Densité")
plt.tight_layout()
plt.savefig("results/residuals_dist.png", dpi=200)
plt.show()
print("✅ residuals_dist.png")

# =========================================================
# 5) MAE glissant (rolling) -> suivi de la précision dans le temps
# =========================================================
def rolling_mae(y, yhat, window):
    diff = np.abs(y - yhat)
    s = pd.Series(diff)
    return s.rolling(window=window, min_periods=max(1, window//4)).mean()

ROLL = max(12, len(pred_df)//50)  # fenêtre par défaut (ajuste)
rm_df = []
for m in model_cols:
    rmae = rolling_mae(pred_df["y"].values, pred_df[m].values, window=ROLL)
    tmp = pd.DataFrame({"ds": pred_df["ds"], "rolling_mae": rmae, "model": m})
    rm_df.append(tmp)
rm_df = pd.concat(rm_df, ignore_index=True)

plt.figure(figsize=(14,4))
sns.lineplot(data=rm_df, x="ds", y="rolling_mae", hue="model")
plt.title(f"MAE glissant (fenêtre={ROLL})")
plt.xlabel("Temps"); plt.ylabel("Rolling MAE")
plt.tight_layout()
plt.savefig("results/rolling_mae.png", dpi=200)
plt.show()
print("✅ rolling_mae.png")

# =========================================================
# 6) Animation (True + prédictions qui se déroulent)
#    -> nécessite ffmpeg installé (sinon, enregistre en .gif via PillowWriter)
# =========================================================
ani_models = model_cols[:]  # choisis quels modèles animer
if len(ani_models) > 0 and len(pred_df) > 5:
    anim_df = pred_df.tail(min(5*H, len(pred_df))).reset_index(drop=True)  # garder une fenêtre raisonnable
    fig, ax = plt.subplots(figsize=(14,4))
    ax.set_title("Animation – True vs Prédictions")
    ax.set_xlabel("Temps"); ax.set_ylabel("y")

    # courbe vraie complète (fine) + point courant
    ax.plot(anim_df["ds"], anim_df["y"], color="black", lw=1, alpha=.4, label="True (historique)")
    true_line, = ax.plot([], [], color="black", lw=2, label="True")

    pred_lines = {m: ax.plot([], [], lw=2, alpha=.9, label=m)[0] for m in ani_models}
    ax.legend(ncol=min(4, len(ani_models)+1))
    plt.xticks(rotation=30)
    plt.tight_layout()

    def init():
        true_line.set_data([], [])
        for m in ani_models:
            pred_lines[m].set_data([], [])
        return [true_line, *pred_lines.values()]

    def update(frame):
        # frame = index jusqu'où on trace
        x = anim_df["ds"].iloc[:frame]
        y = anim_df["y"].iloc[:frame]
        true_line.set_data(x, y)
        for m in ani_models:
            pred_lines[m].set_data(x, anim_df[m].iloc[:frame])
        ax.relim(); ax.autoscale_view()
        return [true_line, *pred_lines.values()]

    frames = len(anim_df)
    anim = FuncAnimation(fig, update, init_func=init, frames=frames, interval=40, blit=False)
    try:
        writer = FFMpegWriter(fps=25, metadata=dict(artist='NeuralForecast'), bitrate=1800)
        anim.save("results/anim_forecast.mp4", writer=writer)
        print("✅ anim_forecast.mp4")
    except Exception as e:
        print("⚠️ ffmpeg indisponible, tentative en GIF…", e)
        from matplotlib.animation import PillowWriter
        anim.save("results/anim_forecast.gif", writer=PillowWriter(fps=15))
        print("✅ anim_forecast.gif")

print("\n🎉 Visuels prêts dans le dossier 'results/'")
```
